```python
import pandas as pd

date_columns = ['order_purchase_timestamp', 'order_approved_at', 'order_delivered_carrier_date', 'order_delivered_customer_date', 'order_estimated_delivery_date', 'shipping_limit_date', 'review_creation_date', 'review_answer_timestamp']

for column in df.columns:
    if column in date_columns:
        try:
            df[column] = pd.to_datetime(df[column])
        except Exception as e:
            print(f"Error converting column {column} to datetime: {e}")

print(df.dtypes)
```

Product analysis

Top selling product

```python
best_selling_product = df.groupby("product_id")["order_item_id"].count().reset_index()

# Rename column for clarity
best_selling_product = best_selling_product.rename(columns={"order_item_id": "total_sold"})

# Get the product with the highest sales
best_selling_product = best_selling_product.sort_values(by="total_sold", ascending=False)

# Display the top-selling product
top_product = best_selling_product.iloc[0]
print(top_product)
```

```python
top_10_products = best_selling_product.head(10)
plt.figure(figsize=(12, 6))
sns.barplot(x="total_sold", y="product_id", data=top_10_products, palette="viridis")

plt.xlabel("Total Sold")
plt.ylabel("Product ID")
plt.title("Top 10 Best-Selling Products")
plt.show()
```

```python
# Count total sales per category
category_sales = df.groupby("product_category_name_english")["order_id"].count().reset_index()

# Rename column for clarity
category_sales = category_sales.rename(columns={"order_id": "total_sold"})

# Sort by most sold category
category_sales = category_sales.sort_values(by="total_sold", ascending=False)
top_category = category_sales.iloc[0]
print(top_category)
```

```python
# Select the top 10 categories
top_10_categories = category_sales.head(10)

# Plot the top 10 selling product categories
plt.figure(figsize=(12, 6))
sns.barplot(x="total_sold", y="product_category_name_english", data=top_10_categories, palette="magma")

plt.xlabel("Total Sold")
plt.ylabel("Product Category")
plt.title("Top 10 Best-Selling Product Categories")
plt.show()
```

Product and Revenue insights

```python
# ----- 1. Calculate Average Price per Product -----
avg_price_per_product = df.groupby("product_id")["price"].mean().reset_index()
avg_price_per_product = avg_price_per_product.rename(columns={"price": "avg_price"})

# ----- 2. Compare Average Price Across Categories -----
avg_price_per_category = df.groupby("product_category_name_english")["price"].mean().reset_index()
avg_price_per_category = avg_price_per_category.rename(columns={"price": "avg_price"})

# Sort categories by average price
avg_price_per_category = avg_price_per_category.sort_values(by="avg_price", ascending=False)

# ----- 3. Identify High-Revenue vs. Low-Revenue Products -----
df["revenue"] = df["price"] * df["order_item_id"]  # Assuming order_item_id represents quantity sold

revenue_per_product = df.groupby("product_id")["revenue"].sum().reset_index()
revenue_per_product = revenue_per_product.sort_values(by="revenue", ascending=False)

top_10_high_revenue = revenue_per_product.head(10)

bottom_10_low_revenue = revenue_per_product.tail(10)


plt.figure(figsize=(12, 6))
sns.barplot(x="avg_price", y="product_category_name_english", data=avg_price_per_category.head(10), palette="coolwarm")
plt.xlabel("Average Price")
plt.ylabel("Product Category")
plt.title("Top 10 Expensive Product Categories")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="revenue", y="product_id", data=top_10_high_revenue, palette="magma")
plt.xlabel("Revenue")
plt.ylabel("Product ID")
plt.title("Top 10 High-Revenue Products")
plt.show()
```

Availability and stock trends

```python
# Convert order purchase timestamp to datetime format
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])

# Extract month & year to analyze trends over time
df["order_month"] = df["order_purchase_timestamp"].dt.to_period("M")

# ----- 1. Analyze Product Availability Trends -----
availability_trends = df.groupby(["order_month", "product_id"])["order_id"].count().reset_index()
availability_trends = availability_trends.rename(columns={"order_id": "total_orders"})

# Plot product availability trends over time (Top 5 products)
top_5_products = df["product_id"].value_counts().index[:5]

plt.figure(figsize=(12, 6))
for product in top_5_products:
    product_trend = availability_trends[availability_trends["product_id"] == product]
    plt.plot(product_trend["order_month"].astype(str), product_trend["total_orders"], marker="o", label=product)

plt.xticks(rotation=45)
plt.xlabel("Month")
plt.ylabel("Total Orders")
plt.title("Product Availability Trends Over Time (Top 5 Products)")
plt.legend()
plt.show()

# ----- 2. Find Out-of-Stock Products Based on Order Rejections -----
out_of_stock_orders = df[df["order_status"].isin(["canceled", "unavailable"])]

out_of_stock_counts = out_of_stock_orders["product_id"].value_counts().reset_index()
out_of_stock_counts.columns = ["product_id", "out_of_stock_count"]

# Top 10 frequently out-of-stock products
top_10_out_of_stock = out_of_stock_counts.head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x="out_of_stock_count", y="product_id", data=top_10_out_of_stock, palette="Reds_r")
plt.xlabel("Number of Out-of-Stock Occurrences")
plt.ylabel("Product ID")
plt.title("Top 10 Frequently Out-of-Stock Products")
plt.show()
```

Average Delivery Time Per Product & Its Impact on Sales

```python
# Convert timestamps to datetime format
df["order_purchase_timestamp"] = pd.to_datetime(df["order_purchase_timestamp"])
df["order_delivered_customer_date"] = pd.to_datetime(df["order_delivered_customer_date"])

# Calculate delivery time (in days)
df["delivery_time"] = (df["order_delivered_customer_date"] - df["order_purchase_timestamp"]).dt.days

# ----- 1. Calculate Average Delivery Time Per Product -----
avg_delivery_time_per_product = df.groupby("product_id")["delivery_time"].mean().reset_index()
avg_delivery_time_per_product = avg_delivery_time_per_product.rename(columns={"delivery_time": "avg_delivery_time"})

# ----- 2. Check the Impact of Longer Delivery Times on Sales -----
sales_per_product = df.groupby("product_id")["order_id"].count().reset_index()
sales_per_product = sales_per_product.rename(columns={"order_id": "total_sales"})

# Merge sales & delivery time data
delivery_vs_sales = avg_delivery_time_per_product.merge(sales_per_product, on="product_id")

# Plot: Relationship Between Delivery Time & Sales
plt.figure(figsize=(10, 5))
sns.scatterplot(x=delivery_vs_sales["avg_delivery_time"], y=delivery_vs_sales["total_sales"], alpha=0.6)
plt.xlabel("Average Delivery Time (Days)")
plt.ylabel("Total Sales")
plt.title("Impact of Delivery Time on Sales")
plt.show()
```
