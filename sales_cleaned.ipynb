```python
# Time between purchases
# First, get all orders with timestamps
customer_purchase_dates = df[['customer_unique_id', 'order_id', 'order_purchase_timestamp']].drop_duplicates()

# Sort by customer and timestamp
customer_purchase_dates = customer_purchase_dates.sort_values(['customer_unique_id', 'order_purchase_timestamp'])

# Calculate time difference between consecutive orders for the same customer
customer_purchase_dates['prev_order_date'] = customer_purchase_dates.groupby('customer_unique_id')['order_purchase_timestamp'].shift(1)
customer_purchase_dates['days_between_orders'] = (customer_purchase_dates['order_purchase_timestamp'] - customer_purchase_dates['prev_order_date']).dt.days

# Filter out first purchases (which have NaN days_between_orders)
repurchase_data = customer_purchase_dates.dropna(subset=['days_between_orders'])

# Calculate average time between purchases
avg_days_between_orders = repurchase_data['days_between_orders'].mean()
median_days_between_orders = repurchase_data['days_between_orders'].median()

print(f"Average days between purchases: {avg_days_between_orders:.2f}")
print(f"Median days between purchases: {median_days_between_orders:.2f}")
```

```python
# 4. Product Association Analysis
print("Analyzing product associations...")

# Create a dataframe with order_id and product_category
order_products = df[['order_id', 'product_category_name_english']].drop_duplicates()

# Find orders with multiple products
order_counts = order_products.groupby('order_id').size().reset_index(name='product_count')
multi_product_orders = order_counts[order_counts['product_count'] > 1]['order_id'].tolist()

# Filter for orders with multiple products
multi_product_df = order_products[order_products['order_id'].isin(multi_product_orders)]

# Create a pivot table for co-occurrence
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# Group by order_id and create a list of products for each order
basket = multi_product_df.groupby('order_id')['product_category_name_english'].apply(list).reset_index()

# Convert to transaction format
te = TransactionEncoder()
te_ary = te.fit(basket['product_category_name_english']).transform(basket['product_category_name_english'])
df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# Apply apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(df_encoded, min_support=0.01, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# Sort by lift
rules = rules.sort_values('lift', ascending=False)

# Display top 10 rules
print("Top 10 Product Association Rules:")
if len(rules) > 0:
    pd.set_option('display.max_columns', None)
    print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
else:
    print("No significant association rules found with the current thresholds.")
```

```python
# 5. Canceled/Returned Orders Analysis
print("Analyzing customer behavior after canceled/returned orders...")

# Identify customers with canceled or returned orders
problem_statuses = ['canceled', 'unavailable']
customers_with_problems = df[df['order_status'].isin(problem_statuses)]['customer_unique_id'].unique()
print(f"Number of customers with canceled/returned orders: {len(customers_with_problems)}")

# For each of these customers, check if they made subsequent purchases
customer_order_dates = df[['customer_unique_id', 'order_id', 'order_purchase_timestamp', 'order_status']].drop_duplicates()
customer_order_dates = customer_order_dates.sort_values(['customer_unique_id', 'order_purchase_timestamp'])

# Create a dataframe to track first problem order and subsequent orders
customer_problem_analysis = []

for customer in customers_with_problems:
    customer_orders = customer_order_dates[customer_order_dates['customer_unique_id'] == customer].copy()
    
    # Find the first problematic order
    problem_orders = customer_orders[customer_orders['order_status'].isin(problem_statuses)]
    if len(problem_orders) == 0:
        continue
        
    first_problem_date = problem_orders['order_purchase_timestamp'].min()
    
    # Check for orders after the first problem
    later_orders = customer_orders[customer_orders['order_purchase_timestamp'] > first_problem_date]
    
    if len(later_orders) > 0:
        days_to_return = (later_orders['order_purchase_timestamp'].min() - first_problem_date).days
        customer_problem_analysis.append({
            'customer_id': customer,
            'returned': True,
            'days_to_return': days_to_return
        })
    else:
        customer_problem_analysis.append({
            'customer_id': customer,
            'returned': False,
            'days_to_return': None
        })

# Convert to dataframe
problem_customer_df = pd.DataFrame(customer_problem_analysis)
```

```python
# 6. RFM Analysis for Customer Segmentation
print("Performing RFM Analysis for Customer Segmentation...")

# Get the most recent order date
max_date = df['order_purchase_timestamp'].max()

# Group by customer and calculate RFM metrics
rfm = df.groupby('customer_unique_id').agg({
    'order_purchase_timestamp': lambda x: (max_date - x.max()).days,  # Recency
    'order_id': 'nunique',  # Frequency
    'payment_value': 'sum'  # Monetary
}).reset_index()

# Rename columns
rfm.columns = ['customer_unique_id', 'recency', 'frequency', 'monetary']
```

```python
print(rfm)
```

```python
# Create quartiles for each metric
rfm['r_quartile'] = pd.qcut(rfm['recency'], 4, labels=False, duplicates='drop')
rfm['f_quartile'] = pd.qcut(rfm['frequency'], 4, labels=False, duplicates='drop')
rfm['m_quartile'] = pd.qcut(rfm['monetary'], 4, labels=False, duplicates='drop')

# Reverse recency (lower is better)
rfm['r_quartile'] = 3 - rfm['r_quartile']

# Calculate RFM score
rfm['rfm_score'] = rfm['r_quartile'] + rfm['f_quartile'] + rfm['m_quartile']

# Create RFM segments
def rfm_segment(df):
    if df['rfm_score'] >= 8:
        return 'Champions'
    elif df['rfm_score'] >= 6:
        return 'Loyal Customers'
    elif df['rfm_score'] >= 4:
        return 'Potential Loyalists'
    elif df['rfm_score'] >= 2:
        return 'At Risk Customers'
    else:
        return 'Lost Customers'

rfm['segment'] = rfm.apply(rfm_segment, axis=1)
```

```python
# Display segment distribution
segment_counts = rfm['segment'].value_counts().reset_index()
segment_counts.columns = ['Segment', 'Count']
segment_percent = segment_counts['Count'] / segment_counts['Count'].sum() * 100
segment_counts['Percent'] = segment_percent.round(2)

print("\nCustomer Segments Distribution:")
print(segment_counts)

# Plot segment distribution
plt.figure(figsize=(12, 6))
sns.barplot(x='Segment', y='Count', data=segment_counts)
plt.title('Customer Segments Distribution')
plt.xlabel('Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
for i, v in enumerate(segment_counts['Count']):
    plt.text(i, v + 50, f"{segment_counts['Percent'][i]}%", ha='center')
plt.tight_layout()
plt.show()
```

```python
# 7. Cohort Analysis
print("Performing Cohort Analysis...")

# Create a copy of the dataframe with only the necessary columns
cohort_data = df[['customer_unique_id', 'order_id', 'order_purchase_timestamp']].drop_duplicates()

# Extract the month of first purchase for each customer (cohort)
cohort_data['cohort_month'] = cohort_data.groupby('customer_unique_id')['order_purchase_timestamp'].transform('min').dt.to_period('M')

# Extract the month of each purchase
cohort_data['purchase_month'] = cohort_data['order_purchase_timestamp'].dt.to_period('M')

# Calculate the month offset (period)
cohort_data['period'] = (cohort_data['purchase_month'] - cohort_data['cohort_month']).apply(lambda x: x.n)
```

```python
# Count unique customers by cohort and period
cohort_counts = cohort_data.groupby(['cohort_month', 'period']).agg(n_customers=('customer_unique_id', 'nunique')).reset_index()

# Get the total number of customers in each cohort
cohort_sizes = cohort_data.groupby('cohort_month').agg(cohort_size=('customer_unique_id', 'nunique')).reset_index()

# Merge the cohort counts with cohort sizes
cohort_data = pd.merge(cohort_counts, cohort_sizes, on='cohort_month')

# Calculate retention rate
cohort_data['retention_rate'] = cohort_data['n_customers'] / cohort_data['cohort_size']

# Create a pivot table for visualization
cohort_pivot = cohort_data.pivot_table(index='cohort_month', columns='period', values='retention_rate')
```

```python
# Plot the cohort heatmap
plt.figure(figsize=(15, 8))
sns.heatmap(cohort_pivot, annot=True, fmt='.0%', cmap='YlGnBu', vmin=0, vmax=0.3)
plt.title('Customer Cohort Analysis: Retention Rates')
plt.xlabel('Month Since First Purchase')
plt.ylabel('Cohort (First Purchase Month)')
plt.tight_layout()
plt.show()
```

```python
# Calculate average retention rates by period
avg_retention = cohort_data.groupby('period')['retention_rate'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='period', y='retention_rate', data=avg_retention, marker='o')
plt.title('Average Retention Rate by Month Since First Purchase')
plt.xlabel('Month Since First Purchase')
plt.ylabel('Average Retention Rate')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
